```python
from statistics import *
# lista de calificaciones
grupoa=[70,70,65,67,75,100,20,100,96]
print(grupoa)
type(grupoa)
```

```python
#sacar la media de todas las calificaciones
prom= sum(grupoa) / len(grupoa)
print(prom)
```

```python
#sacar el promedio mediante la funcion mean
prom2 = mean(grupoa)
print(prom2)
```

```python
from statistics import median
mediana = median(grupoa)

print(mediana)
```

```python
#sacar la moda de las calificaciones
modi = mode(grupoa)
print(modi)
```

```python
#sacar la multimoda
modi2 = multimode(grupoa)
print(modi2)
```

```python
#definimos una variable llamada tiempos100 de tipo lista, que contiene
#los tiempos que tardaron los 10 alumnos para correr los 100 metros
tiempos100 = [15.34,52.0,45.5,45.2,28.4,13.5,40.16,40.16,20.45,22.15,22.15]
 #deplegar la lista
print(tiempos100)
```

```python
#promedio de tiempo usando mean
print("el tiempo promedio de los corredores es => ", mean(tiempos100))
```

```python
#multimoda de 100 metros
print("la moda de los tiempos es => ", multimode(tiempos100))
```

```python
#funcion random
import random
fabrica1 = []
for i in range(100):
  numA = round(random.uniform(50,100),2)
  fabrica1.append(numA)
print(fabrica1)
```

```python
#fabrica 2
fabrica2 = []
for i in range(100):
  numA2 = round(random.uniform(70,100),2)
  fabrica2.append(numA2)
print(fabrica2)
```

```python
#importar python para graficos
import matplotlib.pyplot as plt
```

```python
plt.hist(fabrica1)
plt.title("produccion diaria de la fabrica 1")
plt.show()
```

```python
plt.hist(fabrica2)
plt.title("produccion diaria de la fabrica 2")
plt.show()
```

```python
#media de la fabrica 1
media1 = mean(fabrica1)
mediana1 = median(fabrica1)
moda1 = mode(fabrica1)

print("valores de la fabrica 1")
print("la media es: ",media1)
print("la mediana es: ",mediana1)
print("la moda es: ",moda1)
```

```python
#media de la fabrica 2
media2 = mean(fabrica2)
mediana2 = median(fabrica2)
moda2 = mode(fabrica2)

print("valores de la fabrica 2")
print("la media es: ",media2)
print("la mediana es: ",mediana2)
print("la moda es: ",moda2)
```

```python
#comparacion de graficas
plt.hist(fabrica1)
plt.hist(fabrica2)
plt.title("produccion diaria de las dos fabricas")
plt.show()
```
