```python
import random
import matplotlib.pyplot as plt
from statistics import *
```

```python
#generar 100 numeros aleatorios
import random
fabrica1 = []
for i in range(100):
  num_aleatorio = round(random.uniform(50,100),2)
  fabrica1.append(num_aleatorio)
print(fabrica1)
```

```python
#numeros aleatorios para la fabrica 2
fabrica2=[]
for i in range(100):
  num_aleatorio2 = round(random.uniform(50,100),2)
  fabrica2.append(num_aleatorio2)
print(fabrica2)
```

```python
maximo1= max(fabrica1)
minimo1= min(fabrica1)
rango1= maximo1-minimo1
media1= mean(fabrica1)
mediana1= median(fabrica1)
moda1= mode(fabrica1)
desviacion1= stdev(fabrica1)
print("estadistica descriptiva de la fabrica 1 \n","media= ",media1,"\n","mediana= ",mediana1,"\n","moda= ",moda1,"\n","rango= ",rango1,"\n","desviacion= ",desviacion1)
print("---------------------------------------")
maximo2= max(fabrica2)
minimo2= min(fabrica2)
rango2= maximo2-minimo2
media2= mean(fabrica2)
mediana2= median(fabrica2)
moda2= mode(fabrica2)
desviacion2= stdev(fabrica2)
print("estadistica descriptiva de la fabrica 2 \n","media= ",media2,"\n","mediana= ",mediana2,"\n","moda= ",moda2,"\n","rango= ",rango2,"\n","desviacion= ",desviacion2)
```

```python
plt.figure(figsize=(8,6))
plt.hist(fabrica1,alpha=0.2, bins=10, color='b', edgecolor="black")
plt.axvline(media1, color='b', linestyle='dashed', linewidth=2, label=f'media={media1:.2f}')
plt.axvline(mediana1, color='r', linestyle='dashed', linewidth=2, label=f'mediana={mediana1:.2f}')
plt.axvline(moda1, color='g', linestyle='dashed', linewidth=2, label=f'moda={moda1:.2f}')
plt.axvline(media1-desviacion1, color='y', linestyle='dashed', linewidth=2, label=f'desviacion={desviacion1:.2f}')
plt.axvline(media1+desviacion1, color='y', linestyle='dashed', linewidth=2)
plt.xlabel('valor')
plt.ylabel('frecuencia')
plt.title('produccion diaria de la fabrica 1')
plt.legend()
plt.show()
print("---------------------------------------------------------------------------------------------")
plt.figure(figsize=(8,6))
plt.hist(fabrica2,alpha=0.2, bins=10, color='b', edgecolor="black")
plt.axvline(media2, color='b', linestyle='dashed', linewidth=2, label=f'media={media2:.2f}')
plt.axvline(mediana2, color='r', linestyle='dashed', linewidth=2, label=f'mediana={mediana2:.2f}')
plt.axvline(moda2, color='g', linestyle='dashed', linewidth=2, label=f'moda={moda2:.2f}')
plt.axvline(media2-desviacion2, color='y', linestyle='dashed', linewidth=2, label=f'desviacion={desviacion2:.2f}')
plt.axvline(media2+desviacion2, color='y', linestyle='dashed', linewidth=2)
plt.xlabel('valor')
plt.ylabel('frecuencia')
plt.title('produccion diaria de la fabrica 2')
plt.legend()
plt.show()
```
