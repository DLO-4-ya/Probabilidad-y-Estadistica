```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm
```

```python
plt.figure(figsize=(6,4))

media = 70
desv_std = 4
total_estudiantes = 500
peso_limite = 75
prob_menos_75kg = norm.cdf(peso_limite, media, desv_std)
estudiantes_menos_75kg = total_estudiantes * prob_menos_75kg

print("probabilidad de que un estudiante pese menos de 75 kilogramos: ",estudiantes_menos_75kg)
print("numeri de estudiantes que pean menos de 75 kilogramos: ", estudiantes_menos_75kg)

x = np.linspace(media-4*desv_std, media+4*desv_std, 1000)
y = norm.pdf(x, media, desv_std)

plt.plot(x, y, 'b', linewidth=2)
plt.fill_between(x,y, where=(x <= peso_limite), color='skyblue', alpha=0.5)
plt.title('probabilidad de que un estudiante pese menos de 75 kg. \n $\mu=70$, $\sigma=4$, x=500')
plt.xlabel('peso (kg)')
plt.ylabel('densidad de probabilidad')
plt.grid(True)
plt.show()
```

```python
plt.figure(figsize=(6,4))
media = 70
desv_std=4
total_estudiantes=85
peso_limite=62
prob_menos_62kg=norm.cdf(peso_limite, media, desv_std)
estudiantes_menos_62kg=total_estudiantes*prob_menos_62kg

print("probabilidad de que un estudiante pese menos de 62 kg: ",estudiantes_menos_62kg)
print("numero de estudiantes que pesan menos de 62 kg", estudiantes_menos_62kg)

x=np.linspace(media-4*desv_std, total_estudiantes)
y=norm.pdf(x, media, desv_std)

plt.plot(x,y,'b',linewidth=2)
plt.fill_between(x,y, where=(x<=peso_limite), color='skyblue', alpha=0.5)
plt.title('probabilidad de que un estudiante pese menos de 62 kg. \n $\mu=70$, $\sigma=4$, x=85')
plt.xlabel('peso (kg)')
plt.ylabel('densidad de probabilidad')
```

```python
plt.figure(figsize=(6,4))
media = 70
desv_std=4
total_estudiantes=85
peso_limite=79
prob_menos_79kg=norm.cdf(peso_limite, media, desv_std)
prob_mas_79kg=1-prob_menos_79kg
estudiantes_mas_79kg=total_estudiantes*prob_mas_79kg

print("probabilidad de que un estudiante pese menos de 79 kg: ",prob_mas_79kg)
print("numero de estudiantes que pesan menos de 79 kg", estudiantes_mas_79kg)

x=np.linspace(media-4*desv_std, total_estudiantes)
y=norm.pdf(x, media, desv_std)

plt.plot(x,y,'b',linewidth=2)
plt.fill_between(x,y, where=(x<=peso_limite), color='skyblue', alpha=0.5)
plt.title('probabilidad de que un estudiante pese menos de 79 kg. \n $\mu=70$, $\sigma=4$, x=85')
plt.xlabel('peso (kg)')
plt.ylabel('densidad de probabilidad')
```
