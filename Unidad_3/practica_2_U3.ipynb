```python
import math
import numpy as np
import matplotlib.pyplot as plt
```

```python
x = [5,6,7,8,9,10]
p = [0.05,0.10,0.20,0.30,0.25,0.10]
```

```python
F = np.cumsum(p)

for xi, fi in zip(x,F):
  print(f"F({xi} ={fi:.2f})")
```

```python
plt.step(x,F,where='post')
plt.title("funcion de probabilidad acumulada")
plt.xlabel("calificacion")
plt.ylabel("probabilidad")
plt.grid(True)
plt.ylim(0,1.05)
plt.show()
```
